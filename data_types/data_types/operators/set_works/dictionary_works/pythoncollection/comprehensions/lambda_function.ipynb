```python
# lamda fuction for adding two number
add_number=lambda n1,n2: n1+n2
print(add_number(100,200))
```

```python
#lambda function for substracting two number
sub=lambda n1,n2:n1-n2
print(sub(10,2))
```

```python
#lambda function for max_of_two number
max_two=lambda n1,n2:n1 if n1>n2 else n2
print(max_two(100,200)) 
```

```python
#lambda function for min_of_two number
min_two=lambda n1,n2:n1 if n1<n2 else n2
print(min_two(10,20))
```
